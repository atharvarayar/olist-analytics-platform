```python
from pathlib import Path

data_dir = Path(r"C:\Users\rbray\OneDrive\Desktop\data projs\archive (1)")

files = {
    "customers": "olist_customers_dataset.csv",
    "orders": "olist_orders_dataset.csv",
    "order_items": "olist_order_items_dataset.csv",
    "payments": "olist_order_payments_dataset.csv",
    "products": "olist_products_dataset.csv",
    "reviews": "olist_order_reviews_dataset.csv",
    "sellers": "olist_sellers_dataset.csv",
    "geolocation": "olist_geolocation_dataset.csv",
    "category_translation": "product_category_name_translation.csv",
}
```

```python
def load_table(table_name, filename):
    filepath = data_dir / filename

    print(f"Loading {table_name}...")

    df = pd.read_csv(filepath)

    df.to_sql(
        name=table_name,
        con=engine,
        schema="raw",
        if_exists="replace",
        index=False
    )

    print(f"Loaded {len(df):,} rows into raw.{table_name}")

    return df
```

```python
def load_table(table_name,file_name):
    filepath=data_dir/file_name
    print(f"Loading {table_name}..")
    df=pd.read_csv(filepath)
    df.to_sql(
        name=table_name,
        con=engine,
        schema="raw",
        if_exists="replace",
        index=False
    )
    print(f"Loaded {len(df):,} rows into raw.{table_name}")

    return df
```

```python
loaded_tables={}

for table_name,file_name in files.items():
    loaded_tables[table_name]=load_table(table_name,file_name)
```
